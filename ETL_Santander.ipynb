EXTRACT

```python
import pandas as pd

aluno = pd.read_csv('crossfit.csv', sep=';')
aluno_id = aluno['userID'].tolist()
print(aluno_id)
```

```python
import requests
import json

def get_user(id):
  user_csv = aluno[aluno['userID'] == id]
  if not user_csv.empty:
    return user_csv.iloc[0].to_dict()
  return None

users = [user for id in aluno_id if (user := get_user(id)) is not None]
print(json.dumps(users, indent=2))
```

TRANSFORM


```python
for user in users:
  if user['nivel'] == 'RX':
    user['OBS'] = 'Esse tem futuro!'
  elif user['nivel'] == 'MEDIO':
    user['OBS'] = 'Treine mais.'
  elif user['nivel'] == 'INICIANTE':
    user['OBS'] = 'Mais rapidoooooo!'

print(json.dumps(users, indent=2))
```

LOAD

```python
def update_user(user):
  # Simulating a successful update for local demonstration.
  print(f"Simulating update for user ID {user['userID']} with data: {user}")
  return True # Simulate success

for user in users:
  success = update_user(user)
  if success:
    print(f'user {user["userID"]} atualizado')
  else:
    print(f'Falha ao atualizar user {user["userID"]}')
```

```python
import pandas as pd
updated_aluno_df = pd.DataFrame(users)
updated_aluno_df.to_csv('crossfit.csv', sep=';', index=False)

print("Conteúdo da planilha atualizada:")
display(pd.read_csv('crossfit.csv', sep=';'))
```
